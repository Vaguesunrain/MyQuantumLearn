We have used the vector tools to describe algorithms for a long time. It works well but only if system is in a pure state. If we introduce some noise, or if the system decoheres, how can we describe it? Answer is by using a density matrix.

## 1.From Pure State to Mixed State
If the system is in a pure state $|a\rangle$, its corresponding density matrix is simply:
$$\rho = \vert a\rangle \langle a \vert$$
If we take a convex combination of pure states with different probabilities $p_k$, the mixed system() can be describe as:
$$\rho = \sum_{k = 0}^{m-1} p_k \vert \psi_k\rangle \langle \psi_k \vert,$$


## 2.Density matrices and the spectral theorem
We have seen that if we take a convex combination of pure states，we get a density matrix:
$$\rho = \sum_{k = 0}^{m-1} p_k \vert \psi_k\rangle \langle \psi_k \vert$$
However, a density matrix can be composited by several different state sets with corresponding probability, there exists an orthonormal basis of n dimensional complex vectors to rebuild it. We can find the basis through spectral theorem:

Suppose that $M$ is an n-dimension matrix, there is the orthonormal basis $\{\vert\psi_0\rangle,\ldots,\vert\psi_{n-1}\rangle \}$ with distinct complex numbers $\lambda_0,\ldots,\lambda_{n-1}$ satisfing that:
$$M = \lambda_0 \vert \psi_0\rangle\langle \psi_0\vert + \cdots + \lambda_{n-1} \vert \psi_{n-1}\rangle\langle \psi_{n-1}\vert.$$
According to Cyclic properity of trace: $tr(AB)=tr(BA)$, we have:$\operatorname{Tr}\bigl(\vert\phi\rangle\langle\phi\vert\bigr)
= \operatorname{Tr}\bigl(\langle\phi\vert\phi\rangle\bigr) = \langle\phi\vert\phi\rangle.$
So, we obtain a conlusion reveals that sum of probability equals 1:
$$\begin{gathered}
1 = \operatorname{Tr}(\rho) =
\operatorname{Tr}\bigl(\lambda_0 \vert \psi_0\rangle\langle \psi_0\vert + \cdots + \lambda_{n-1} \vert \psi_{n-1}\rangle\langle \psi_{n-1}\vert\bigr)\\[2mm]
= \lambda_0 \operatorname{Tr}\bigl(\vert \psi_0\rangle\langle \psi_0\vert\bigr) + \cdots + \lambda_{n-1} \operatorname{Tr}\bigl(\vert \psi_{n-1}\rangle\langle \psi_{n-1}\vert\bigr)
= \lambda_0 + \cdots + \lambda_{n-1}
\end{gathered}$$

Here is an example, let's focus on a matrix combined by two non-orthogonal superposition states :
$$\frac{1}{2} \vert 0\rangle\langle 0 \vert + \frac{1}{2} \vert +\rangle\langle + \vert
= \begin{pmatrix}
\frac{3}{4} & \frac{1}{4}\\[2mm]
\frac{1}{4} & \frac{1}{4}
\end{pmatrix}
= \cos^2(\pi/8) \vert \psi_{\pi/8} \rangle \langle \psi_{\pi/8}\vert +
\sin^2(\pi/8) \vert \psi_{5\pi/8} \rangle \langle \psi_{5\pi/8}\vert$$

## 3.Reduced states and the partial trace
Before we start to learn Reduced states and the partial trace, let us consider a simple example of a composite system composed of two subsystems, X and Y:
\begin{array}{c|cc|cc} 
 & |0\rangle_X|0\rangle_Y & |0\rangle_X|1\rangle_Y & |1\rangle_X|0\rangle_Y & |1\rangle_X|1\rangle_Y \\ \hline
\langle 0|_X\langle 0|_Y & \rho_{00,00} & \rho_{00,01} & \rho_{00,10} & \rho_{00,11} \\
\langle 0|_X\langle 1|_Y & \rho_{01,00} & \rho_{01,01} & \rho_{01,10} & \rho_{01,11} \\ \hline
\langle 1|_X\langle 0|_Y & \rho_{10,00} & \rho_{10,01} & \rho_{10,10} & \rho_{10,11} \\
\langle 1|_X\langle 1|_Y & \rho_{11,00} & \rho_{11,01} & \rho_{11,10} & \rho_{11,11}
\end{array}

The mostly important is the values in diagonal which means the probabilitys we can obtain the relative states when we measure(For example, we can obtain the state $|00\rangle$ with the probability $\rho_{00,00}$). The values in non-diagonal means coherence between all kinds of states.

For simplicity, we begin with the example: reduced states for an e-bit.
Qubits A and B are entangled, and Alice holds A, Bob holds B. Our purpose is to figure out how to describe the qubit A independently. Assume that system (A,B) is in the Bell state, $\vert\phi^+\rangle = \frac{1}{\sqrt{2}} \vert 00 \rangle + \frac{1}{\sqrt{2}} \vert 11 \rangle.$
Now, Bob thinks that if I measure my own qubit, it is reasonable to derive the state via the probability. So, he apply projective measurement to his qubit:
$$\bigl\| \bigl( \mathbb{I}_{\mathsf{A}} \otimes \langle 0\vert \bigr) \vert \phi^+ \rangle \bigr\|^2
= \Bigl\| \frac{1}{\sqrt{2}} \vert 0 \rangle \Bigr\|^2 = \frac{1}{2},$$
$$\bigl\| \bigl( \mathbb{I}_{\mathsf{A}} \otimes \langle 1\vert \bigr) \vert \phi^+ \rangle \bigr\|^2
= \Bigl\| \frac{1}{\sqrt{2}} \vert 1 \rangle \Bigr\|^2 = \frac{1}{2}$$
As Bob gets the result, we can know that the matrix about Alice or Bob can be described as:
$$\frac{1}{2} \vert 0\rangle\langle 0\vert + \frac{1}{2} \vert 1\rangle\langle 1\vert = \frac{1}{2} \mathbb{I}.$$
And Alice's system can be written as:$\frac{1}{2} \vert 0\rangle\langle 0\vert + \frac{1}{2} \vert 1\rangle\langle 1\vert = \frac{1}{2} \mathbb{I}_{\mathsf{A}},$ which has nothing related with Bob.

However, the main thought is not telling us get system of Alice by know Bob's firstly. It actully tells us reduced state of Alice has nothing related with Bob.

### 3.1 Further derivation
Now, let us generalize this to arbitrary systems A and B with classical state sets $\Sigma$ and $\Gamma$, respectively. A density matrix $\rho$ representing a state of the combined system (A, B) has row and column indices corresponding to the Cartesian product $\Sigma \times \Gamma$.Suppose the state of (A, B) is described by a pure quantum state vector $\vert\psi\rangle$. We can decompose the composite system by expanding it in terms of the basis vectors of B:$$\vert\psi\rangle = \sum_{b\in\Gamma} \vert\phi_b\rangle \otimes \vert b\rangle.$$ This form means that we disperse the composited system by the basis of B. Then we extract the part of B and do nothing on A. So we can get the part of A:$|\phi_b\rangle = (\mathbb{I}_A \otimes \langle b|) |\psi\rangle$.
Here, we project out the component corresponding to system B while leaving system A unaltered. This yields the unnormalized state vector for system A:$$|\phi_b\rangle = (\mathbb{I}_A \otimes \langle b|) |\psi\rangle.$$If we perform a standard basis measurement on B, we obtain outcome $b$ with probability $P(b) = \||\phi_b\rangle\|^2$. Concurrently, the state of A conditioned on this outcome becomes the normalized vector:$$\frac{\vert \phi_b \rangle}{\|\vert\phi_b\rangle\|}.$$By summing over all possible measurement outcomes weighted by their respective probabilities, the overall density matrix of A can be expressed as:
$$\rho_{\mathsf{A}} = \sum_{b\in\Gamma}
\|\vert\phi_b\rangle\|^2 \frac{\vert \phi_b \rangle\langle\phi_b\vert}{\|\vert\phi_b\rangle\|^2}
= \sum_{b\in\Gamma} \vert \phi_b \rangle\langle\phi_b\vert
= \sum_{b\in\Gamma}
\bigl(\mathbb{I}_{\mathsf{A}} \otimes \langle b\vert\bigr) \vert\psi\rangle\langle\psi\vert
\bigl(\mathbb{I}_{\mathsf{A}} \otimes \vert b\rangle\bigr).$$


### 3.2 The partial trace
To obtain the reduced state or reduced density matrix of A, the operation applied to the composite system is denoted by partial trace. In the example above, we trace out B:
$$\operatorname{Tr}_{\mathsf{B}} (\rho) =
\sum_{b\in\Gamma}
\bigl( \mathbb{I}_{\mathsf{A}} \otimes \langle b \vert\bigr)
\rho
\bigl( \mathbb{I}_{\mathsf{A}} \otimes \vert b \rangle\bigr).$$
Correspondingly, if we want to get reduced density matrix of B, we can trace out A:
$$\operatorname{Tr}_{\mathsf{A}} (\rho) =
\sum_{a\in\Sigma}
\bigl(\langle a \vert\otimes\mathbb{I}_{\mathsf{B}}\bigr)
\rho
\bigl(\vert a \rangle\otimes\mathbb{I}_{\mathsf{B}}\bigr).$$ 
For a more complex system (A,B,C), we also have:
$$\begin{aligned}
\rho_{\mathsf{AC}} & = \operatorname{Tr}_{\mathsf{B}}(\rho) = \sum_{b\in\Gamma}
\bigl( \mathbb{I}_{\mathsf{A}} \otimes \langle b \vert \otimes \mathbb{I}_{\mathsf{C}} \bigr) \rho
\bigl( \mathbb{I}_{\mathsf{A}} \otimes \vert b \rangle \otimes \mathbb{I}_{\mathsf{C}} \bigr) \\[2mm]
\rho_{\mathsf{C}} & = \operatorname{Tr}_{\mathsf{AB}}(\rho) = \sum_{a\in\Sigma} \sum_{b\in\Gamma}
\bigl( \langle a \vert \otimes \langle b \vert \otimes \mathbb{I}_{\mathsf{C}} \bigr) \rho
\bigl( \vert a \rangle \otimes \vert b \rangle \otimes \mathbb{I}_{\mathsf{C}} \bigr)
\end{aligned}$$

#### 3.2.1 Operational Rules
By the way, I will introduce how it works when we have only the forms of matrices. We describe $Tr_A$ and $Tr_B$ as:
$$\begin{aligned}
\operatorname{Tr}_{\mathsf{A}}(M \otimes N) & = \operatorname{Tr}(M) N \\[2mm]
\operatorname{Tr}_{\mathsf{B}}(M \otimes N) & = \operatorname{Tr}(N) M.
\end{aligned}$$
For a three-qubit system (A,B,C),  to trace out subsystems A and C,we can perform the operation iteratively:
$$\operatorname{Tr}_{\mathsf{AC}}(\rho_{\mathsf{ABC}}) = \operatorname{Tr}_{\mathsf{A}}\bigl(\operatorname{Tr}_{\mathsf{C}}(\rho_{\mathsf{ABC}})\bigr) = \operatorname{Tr}_{\mathsf{C}}\bigl(\operatorname{Tr}_{\mathsf{A}}(\rho_{\mathsf{ABC}})\bigr) = \rho_{\mathsf{B}}$$

This method provides an effective tool when we focus exclusively on a specific subsystem of a larger composite structure. By tracing out the remaining degrees of freedom, it not only simplifies the mathematical representation of the density matrix, but also reminds us that complete global information is not always required to fully characterize local quantum behaviors.
