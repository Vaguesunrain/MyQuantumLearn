# 1. Quantum channel basis
For an unique channel ,we can use $\Phi$ to denote it. Suppose that there is a system with density matrix $\rho$ pass through the channel which can be describe as $\Phi(\rho)$.
Let's focus on two examples. Firstly, if the matrix is convex situation:$p\rho + (1-p)\sigma$. We can know that，
$$\Phi(p\rho + (1-p)\sigma) = p\Phi(\rho) + (1-p)\Phi(\sigma).$$
Secondly, if a system consists of (Z,X) and we hope to transform it into (Z,Y). We can get:
$$\bigl(\Phi\otimes\operatorname{Id}_{\mathsf{Z}}\bigr)(\rho) = \sum_{a,b = 0}^{m-1} \Phi(\rho_{a,b}) \otimes \vert a\rangle\langle b\vert.$$
Above the paragraph explains an unique channel applys into different system, And how about different channels?More generally, if we have channels $\Phi_{0},\ldots,\Phi_{m-1}$ and a probability vector $(p_0,\ldots, p_{m-1})$, then we can average these channels together to obtain a new channel:
$$\sum_{k = 0}^{m-1} p_k \Phi_k.$$
